```python
# prompt: # gere 10 cenarios com o preco medio calculado com o modelo monte carlo com esse vetor de producao prevista para os proximos 10 anos
# # 2025	2026	2027	2028	2029	2030	2031	2032	2033	2034
# # 138,00	191,00	334,00	381,00	435,00	543,00	494,00	452,00	398,00	389,00
# #  This is not code, so there are no syntax errors to correct.
# #  It appears to be a prompt for a Monte Carlo simulation.
# #  To provide Python code, please specify the details of the simulation, including any relevant parameters or distributions.


def monte_carlo_price_forecast(production_data, num_simulations=1000):
    """
    Generates price forecasts using Monte Carlo simulation.

    Args:
        production_data: A pandas Series or list representing the predicted production for each year.
        num_simulations: The number of simulations to run.

    Returns:
        A pandas DataFrame with average price for each year across all simulations.
    """

    # Example: Assume a base price and some random variation based on production
    base_price = 70  # Example base price, adjust as needed
    price_volatility = 0.1  # Example volatility, adjust as needed

    years = range(2025, 2035)
    simulated_prices = pd.DataFrame(index=years, columns=range(num_simulations))


    for year_index, year in enumerate(years):
        for simulation in range(num_simulations):
          # Calculate price variation based on production
          price_variation = np.random.normal(0, price_volatility * production_data[year_index])
          simulated_prices.loc[year, simulation] = base_price * (1+price_variation)

    # Calculate average price across the simulations
    average_prices = simulated_prices.mean(axis=1)

    return average_prices


# Example usage:
production_forecast = [50370000, 69715000, 121910000, 139446000, 158775000, 18031000, 165432000, 145270000, 141985000]
average_prices = monte_carlo_price_forecast(production_forecast)
average_prices
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Remover o estilo seaborn-v0_8-darkgrid

# Obter os dados do histograma (preços simulados no último dia)
simulated_prices_last_day = simulation_df.iloc[-1]

# Calcular o histograma (contagens e bins)
counts, bin_edges = np.histogram(simulated_prices_last_day, bins=15, density=True)

# Calcular a CDF (cumulative distribution function)
cdf = np.cumsum(counts) * np.diff(bin_edges)

# Criar o gráfico da CDF
plt.figure(figsize=(12, 7))
plt.plot(bin_edges[1:], cdf, marker='.', linestyle='-', markersize=8)

# Formatar o eixo y para mostrar porcentagens
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.0%}'.format(y)))

# Calcular os percentis 20 e 47.75
percentile_20 = np.percentile(simulated_prices_last_day, 20)
percentile_47_75 = np.percentile(simulated_prices_last_day, 47.75)


# Adicionar marcadores para os percentis 20% e 47.75%
plt.axvline(percentile_20, color='purple', linestyle='--', linewidth=1, label=f'20º Percentil: {percentile_20:.2f}')
plt.axvline(percentile_47_75, color='orange', linestyle='--', linewidth=1, label=f'47.75º Percentil: {percentile_47_75:.2f}')

# Adicionar anotações para os percentis
plt.text(percentile_20, 0.25, f'{percentile_20:.2f}', color='purple', ha='right')
plt.text(percentile_47_75, 0.55, f'{percentile_47_75:.2f}', color='orange', ha='left')

# Destacar a área à esquerda de 47.75% com uma sombra
plt.fill_between(bin_edges[1:], 0, cdf, where=(bin_edges[1:] <= percentile_47_75), color='orange', alpha=0.3, label='Área abaixo de 47.75%')


plt.xlabel('Preço Simulado')
plt.ylabel('Percentual Acumulado')
plt.title('Distribuição Cumulativa dos Preços Simulados')
plt.legend()
plt.show()
```
